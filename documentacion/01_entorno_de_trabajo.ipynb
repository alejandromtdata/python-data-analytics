# 01. Entorno de trabajo

Antes de comenzar un proyecto de análisis de datos con Python, necesitamos preparar un entorno de trabajo que nos permita organizar el código, instalar las librerías necesarias y ejecutar nuestros análisis.

En este documento veremos cómo:

1. Crear un repositorio a partir de una plantilla de GitHub.
2. Clonar el repositorio en nuestro equipo.

4. Instalar las librerías necesarias.

6. Comprobar que todo funciona correctamente.

## 1. Crear un proyecto a partir de una plantilla de GitHub

Una forma práctica de comenzar un proyecto nuevo es utilizar una plantilla que ya tenga preparada una estructura básica.

En GitHub:

1. Buscamos nuestro repositorio plantilla: `plantilla_datascience`.
2. Pulsamos **Use this template**.
3. Seleccionamos **Create a new repository**.
4. Introducimos el nombre del nuevo repositorio.
5. Creamos el repositorio.

De esta manera, obtenemos un nuevo repositorio basado en la estructura de nuestra plantilla, sin tener que crear todos los archivos y carpetas desde cero.

## 2. Clonar el repositorio

Una vez creado el repositorio en GitHub, necesitamos descargar una copia local para trabajar desde nuestro equipo.

Para ello utilizamos `git clone`.

```bash
git clone https://github.com/usuario/nombre_del_repo.git

## 3. Preparación del entorno Python

En un proyecto de análisis de datos utilizaremos diferentes herramientas y librerías.

Entre las principales encontramos:

- Python.
- Pandas.
- NumPy.
- Matplotlib.
- Seaborn.
- Jupyter.

Para trabajar correctamente necesitamos un entorno independiente para cada proyecto.

### ¿Qué es un entorno virtual?

Un entorno virtual es un espacio aislado que permite instalar y gestionar las dependencias de un proyecto sin interferir con las de otros proyectos o con la instalación global de Python.

Esto evita conflictos entre versiones de librerías y facilita reproducir el entorno de trabajo.

### Gestión del entorno con uv

En nuestros proyectos utilizaremos principalmente `uv`.

`uv` permite gestionar el proyecto Python, sus dependencias y el entorno virtual.

Para comenzar, abrimos la terminal de VS Code dentro de la carpeta del proyecto.

``` bash 
uv init

Este comando inicializa el proyecto y crea los archivos de configuración necesarios, entre ellos `pyproject.toml`.

A continuación, instalamos las librerías que vamos a utilizar:
```bash 
uv add pandas numpy matplotlib seaborn jupyter ipykernel

Con `uv add` añadimos las dependencias al proyecto y las instalamos en su entorno virtual.

Entre los archivos que podemos encontrar después de preparar el proyecto están:

- `pyproject.toml`: configuración del proyecto y declaración de dependencias.
- `uv.lock`: registra las versiones concretas de las dependencias resueltas.
- `.venv/`: carpeta que contiene el entorno virtual.

**Nota:** la carpeta `.venv` es un entorno local del proyecto. No debemos subirla a GitHub.


### Alternativa: crear un entorno con venv

Python también incluye una herramienta propia para crear entornos virtuales: `venv`.

Podemos utilizarla directamente desde la terminal:

```bash 
python3 -m venv .venv

Este comando crea un entorno virtual llamado `.venv`.

Para activarlo en macOS o Linux:
```bash 
source .venv/bin/activate

Una vez activado, podemos instalar las librerías necesarias utilizando `pip`.

Por ejemplo:

```bash
pip install pandas numpy matplotlib seaborn jupyter ipykernel

**¿uv o venv?**

Ambas herramientas permiten trabajar con entornos virtuales.

- `venv` crea y gestiona entornos virtuales.
- `uv` ofrece una gestión más amplia del proyecto, las dependencias y el entorno.

En nuestros proyectos utilizaremos principalmente `uv`.

## 4. Seleccionar el intérprete de Python en VS Code

VS Code necesita saber qué instalación de Python debe utilizar para ejecutar nuestro código.

Para seleccionar el intérprete:

1. Abrimos la paleta de comandos con `Command + Shift + P`.
2. Buscamos `Python: Select Interpreter`.
3. Seleccionamos el intérprete perteneciente al entorno virtual del proyecto.

Normalmente aparecerá una ruta similar a:

`.venv/bin/python`

Debemos seleccionar el Python del entorno virtual y no una instalación global diferente.

De esta manera, VS Code utilizará las librerías instaladas específicamente para ese proyecto.

## 5. Configurar Jupyter y el kernel

Jupyter permite trabajar mediante Notebooks (`.ipynb`), combinando explicaciones, código Python y resultados en un mismo documento.

### ¿Qué es un kernel?

El kernel es el proceso que ejecuta el código Python de un Notebook.

Cuando ejecutamos una celda, Jupyter envía el código al kernel seleccionado y muestra el resultado.

Es importante seleccionar el kernel correspondiente al entorno virtual del proyecto para que el Notebook pueda utilizar las librerías que hemos instalado.

### Registrar el entorno como kernel

Si necesitamos registrar manualmente nuestro entorno para que aparezca como kernel de Jupyter, podemos ejecutar:

```bash
uv run python -m ipykernel install --user --name nombre_del_proyecto

## 6. Crear y comprobar un Notebook

Para crear un Notebook:

1. Abrimos la carpeta del proyecto en VS Code.
2. Creamos un archivo con extensión `.ipynb` dentro de la carpeta correspondiente.
3. Abrimos el Notebook.
4. Seleccionamos el kernel asociado al entorno virtual del proyecto.

Una vez seleccionado, podemos comprobar que las librerías están disponibles.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Entorno de trabajo preparado correctamente")

Si las importaciones se ejecutan sin errores, significa que las librerías están disponibles en el entorno seleccionado y podemos comenzar a trabajar.

No obstante, esta comprobación no valida por sí sola todo el proyecto: también debemos asegurarnos de que VS Code utiliza el intérprete y el kernel correctos.
