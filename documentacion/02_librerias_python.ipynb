## 2. Librerías de Python

Una librería es un conjunto de herramientas que podemos utilizar desde Python para realizar determinadas tareas sin tener que programarlas desde cero.

Las principales librerías que utilizaremos:

| Librería | ¿Para qué la utilizamos? |
|---|---|
| **NumPy** | Cálculos numéricos y operaciones con datos. |
| **Pandas** | Cargar, limpiar, transformar y analizar datos. |
| **Matplotlib** | Crear visualizaciones. |
| **Seaborn** | Crear visualizaciones estadísticas. |
| **Jupyter** | Trabajar de forma interactiva mediante Notebooks. |
