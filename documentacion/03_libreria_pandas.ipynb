# 3. Librería Pandas

Pandas es una librería de Python utilizada para trabajar con datos de forma estructurada. Permite cargar, organizar, limpiar, transformar y analizar datasets.

## 3.1 Importar Pandas

Para utilizar Pandas en Python, primero debemos importar la librería:

```python
import pandas as pd

## 3.2 DataFrame

Un DataFrame es una estructura de datos de Pandas formada por filas y columnas, similar a una tabla.

```python
ventas = {
    "cliente": ["Ana", "Carlos", "Laura"],
    "producto": ["Monitor", "Teclado", "Ratón"],
    "importe": [500, 240, 250]
}

df_ventas = pd.DataFrame(ventas)

df_ventas
```

## 3.3 Carga de datos

Para cargar un archivo CSV utilizamos `read_csv()`.

```python
df = pd.read_csv("../data/raw/data.csv")

df
```

`df` muestra el DataFrame que acabamos de cargar.

## 3.4 Exploración inicial

Una vez cargados los datos, podemos realizar una primera exploración del DataFrame.

```python
df.head()       # Muestra las 5 primeras filas por defecto.
df.tail()       # Muestra las 5 últimas filas.
df.sample(5)    # Muestra 5 filas aleatorias.

df.shape        # Devuelve una tupla: (filas, columnas).
df.shape[0]     # Devuelve el número de filas.
df.shape[1]     # Devuelve el número de columnas.

df.columns      # Muestra los nombres de las columnas.
df.dtypes       # Muestra los tipos de datos de cada columna.

df.info()       # Información de la tabla: tipos, nulos y memoria.
df.describe()   # Estadísticas básicas de las columnas numéricas.
```

## 3.5 Análisis de columnas

Podemos analizar una columna concreta para conocer sus características.

```python
df["nombre_columna"].size
# Número total de filas (incluye NaN).

df["nombre_columna"].count()
# Número de valores NO nulos.

df["nombre_columna"].nunique()
# Número de valores únicos (ignora NaN).

df["nombre_columna"].value_counts()
# Número de veces que aparece cada valor distinto dentro de la columna.
```

## 3.6 Operaciones sobre columnas

Pandas permite realizar operaciones estadísticas directamente sobre las columnas.

```python
df["edad"].mean()   # Calcula la media.
df["edad"].min()    # Devuelve el valor mínimo.
df["edad"].max()    # Devuelve el valor máximo.
df["edad"].sum()    # Suma los valores.
```

**Nota:** Estas operaciones ignoran los valores `NaN`, se adaptan al tamaño del dataset y devuelven un número.

## 3.7 Filtrado de datos

Podemos filtrar las filas de un DataFrame utilizando condiciones.

```python
df[df["edad"] > 30]
# Devuelve solo las filas donde edad es mayor que 30.

df[df["salario"] > 25000]

df[df["ciudad"] == "Madrid"]
# Devuelve solo las filas donde ciudad tiene el valor "Madrid".
```

## 3.8 Combinación de condiciones

Para combinar condiciones sobre las columnas de un DataFrame utilizamos:

- `&` para AND: deben cumplirse todas las condiciones.
- `|` para OR: debe cumplirse al menos una de las condiciones.

Cada condición debe ir entre paréntesis.

```python
df[(df["edad"] > 30) & (df["salario"] > 25000)]
# Devuelve las filas donde edad es mayor que 30 Y salario es mayor que 25000.

df[(df["ciudad"] == "Madrid") | (df["ciudad"] == "Valencia")]
# Devuelve las filas donde ciudad es Madrid O Valencia.
```

## 3.9 Crear nuevas columnas

Podemos crear una nueva columna a partir de operaciones realizadas sobre otras columnas.

```python
df["salario_mensual"] = df["salario"] / 12
```

Este código crea la columna `salario_mensual` calculando el salario anual dividido entre 12. No modifica los valores de la columna original `salario`.

## 3.10 Localizar valores máximos y mínimos

Podemos obtener el valor máximo o mínimo de una columna y localizar el índice de la fila donde aparece.

```python
df["salario"].max()
# Devuelve el valor máximo de la columna salario.

df["salario"].min()
# Devuelve el valor mínimo de la columna salario.

df["salario"].idxmax()
# Devuelve el índice de la fila donde aparece el salario máximo.

df["salario"].idxmin()
# Devuelve el índice de la fila donde aparece el salario mínimo.
```

Si queremos obtener la fila completa donde se encuentra el valor máximo o mínimo, utilizamos `.loc[]` junto con `idxmax()` o `idxmin()`.

```python
df.loc[df["salario"].idxmax()]
# Devuelve la fila completa con el salario máximo.

df.loc[df["salario"].idxmin()]
# Devuelve la fila completa con el salario mínimo.
```

**Diferencia importante:**

- `max()` y `min()` devuelven el valor máximo y mínimo.
- `idxmax()` y `idxmin()` devuelven el índice de la fila donde se encuentra ese valor.
- `.loc[]` permite recuperar la fila utilizando ese índice.

## 3.11 Comparar valores con una referencia

Podemos calcular un valor de referencia, como la media de una columna, y utilizarlo para filtrar el DataFrame.

```python
media_salario = df["salario"].mean()
# Calcula la media del salario y la guarda en la variable media_salario.

df[df["salario"] > media_salario]
# Devuelve las filas de las personas cuyo salario está por encima de la media.
```

La comparación genera una condición booleana (`True` o `False`) para cada fila. El filtrado devuelve únicamente las filas donde la condición es `True`.

## 3.12 Agrupaciones con groupby()

`groupby()` permite agrupar las filas de un DataFrame según los valores de una o varias columnas y realizar cálculos sobre cada grupo.

```python
df.groupby("ciudad")["salario"].mean()
```

En este ejemplo:

1. `groupby("ciudad")` agrupa las filas por cada ciudad distinta.
2. `["salario"]` selecciona la columna sobre la que queremos realizar el cálculo.
3. `.mean()` calcula la media del salario para cada ciudad.

El resultado muestra la media salarial de cada ciudad.

**Nota sobre los valores nulos:** `mean()` ignora los salarios nulos (`NaN`) al calcular la media. Además, por defecto, `groupby()` no genera un grupo para las filas cuya columna de agrupación (`ciudad`, en este ejemplo) tiene un valor nulo.
