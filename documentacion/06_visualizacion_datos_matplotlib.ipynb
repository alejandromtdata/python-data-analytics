# 06. Visualización de datos con Matplotlib

## 6.1 ¿Qué es Matplotlib?

Matplotlib es una librería de Python utilizada para crear visualizaciones de datos. Permite representar información mediante diferentes tipos de gráficos para facilitar su exploración, comparación y comunicación.

En Data Analytics se utiliza principalmente para:

- Visualizar tendencias.
- Comparar categorías.
- Estudiar distribuciones.
- Analizar relaciones entre variables.

## 6.2 Importación de Matplotlib

Importamos `pyplot`, que proporciona las funciones más utilizadas para crear y personalizar gráficos.

```python
import matplotlib.pyplot as plt
```

El alias habitual es `plt`.

## 6.3 ¿Cómo trabajamos con Matplotlib?

Para crear un gráfico necesitamos los datos que queremos representar y una función que indique qué tipo de gráfico queremos crear.

Los datos pueden proceder de:

- Listas de Python.
- Arrays de NumPy.
- Columnas de un DataFrame de Pandas.

El proceso básico es:

1. Preparar los datos.
2. Elegir el tipo de gráfico.
3. Pasar los datos a Matplotlib.
4. Personalizar el gráfico si es necesario.
5. Mostrar el gráfico.

```python
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo"]
ventas = [1200, 1500, 1100, 1800, 2100]

plt.plot(meses, ventas)
plt.show()
```

- `meses`: datos que representaremos en el eje X.
- `ventas`: datos que representaremos en el eje Y.
- `plt.plot()`: crea un gráfico de líneas.
- `plt.show()`: muestra el gráfico.

## 6.4 Tipos de gráficos principales

### Gráfico de líneas: `plt.plot()`

Muestra la evolución de una variable a lo largo de un eje ordenado. Es especialmente útil cuando existe una dimensión temporal, aunque no es obligatorio que el eje represente el tiempo.

```python
plt.plot(meses, ventas)
plt.show()
```

<img src="../docs/images/ejemplo_grafico_lineas.png" alt="Ejemplo de gráfico de líneas" width="400">

### Gráfico de barras verticales: `plt.bar()`

Permite comparar valores entre diferentes categorías.

```python
categorias = ["A", "B", "C"]
valores = [10, 15, 8]

plt.bar(categorias, valores)
plt.show()
```

<img src="../docs/images/ventas_por_categorias.png" alt="Ventas por categorias" width="400">

### Gráfico de barras horizontales: `plt.barh()`

Muestra la misma información que un gráfico de barras verticales, pero en horizontal.

Es especialmente útil cuando hay muchas categorías o sus nombres son largos, ya que facilita la lectura de las etiquetas.

```python
plt.barh(categorias, valores)
plt.show()
```

<img src="../docs/images/grafico_de_horizontales.png" alt="grafico de barras horizontales" width="400">

### Histograma: `plt.hist()`

Muestra la distribución de una variable numérica agrupando los valores en intervalos.

Permite observar la forma de la distribución, la concentración de valores y posibles valores extremos.

```python
precios = [5, 8, 8, 10, 12, 14, 15, 16, 18, 20]

plt.hist(precios)
plt.show()
```

<img src="../docs/images/histograma.png" alt="histograma" width="400">

### Gráfico de dispersión: `plt.scatter()`

Muestra la relación entre dos variables numéricas. Cada punto representa una observación del dataset con un valor en el eje X y otro en el eje Y.

Permite identificar posibles tendencias positivas, negativas o la ausencia de una relación visual clara.

```python
precio = [5, 8, 12, 15, 20, 25]
ventas = [200, 180, 160, 120, 100, 80]

plt.scatter(precio, ventas)
plt.show()
```

<img src="../docs/images/grafico_de_dispersion.png" alt="Gráfico de dispersión" width="400">

## 6.5 Personalización y configuración de gráficos

En proyectos de análisis de datos es recomendable trabajar con los objetos `fig` y `ax`, ya que permiten controlar y personalizar el gráfico de forma clara y flexible.

### 6.5.1 Figura y ejes: `fig` y `ax`

```python
fig, ax = plt.subplots()

ax.plot(meses, ventas)
ax.set_title("Ventas por mes")
ax.set_xlabel("Meses")
ax.set_ylabel("Ventas")

plt.show()
```

<img src="../docs/images/grafico_con_axes.png" alt="gráfico con axes" width="400">

- `fig`: representa la figura completa.
- `ax`: representa el área donde se dibuja el gráfico.
- `ax.plot()`: crea el gráfico.
- `ax.set_title()`: establece el título.
- `ax.set_xlabel()`: establece la etiqueta del eje X.
- `ax.set_ylabel()`: establece la etiqueta del eje Y.

### 6.5.2 Personalización básica

Podemos modificar el aspecto de una línea mediante diferentes parámetros:

```python
fig, ax = plt.subplots()

ax.plot(
    meses,
    ventas,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=8
)

plt.show()
```

- `color`: color de la línea.
- `linestyle`: estilo de la línea.
- `linewidth`: grosor de la línea.
- `marker`: marcador de cada punto.
- `markersize`: tamaño del marcador.

### 6.5.3 Etiquetas sobre las barras

Podemos añadir el valor de cada barra para facilitar la lectura del gráfico.

```python
valores = [3, 7, 2]

fig, ax = plt.subplots()
ax.bar(range(len(valores)), valores)

for i, v in enumerate(valores):
    ax.text(i, v + 0.1, str(v), ha="center")

plt.show()
```

En `ax.text()`:

- `i`: posición horizontal donde se coloca el texto.
- `v + 0.1`: posición vertical. En este ejemplo, la etiqueta queda ligeramente por encima de la barra.
- `str(v)`: valor que se muestra convertido a texto.
- `ha="center"`: centra la etiqueta horizontalmente. También podemos utilizar `"left"` o `"right"`.

Podemos modificar el texto mostrado. Por ejemplo:

```python
ax.text(i, v + 0.1, f"{v} unidades", ha="center")
```

### 6.5.4 Tamaño del gráfico

El parámetro `figsize` establece el tamaño de la figura en pulgadas: primero el ancho y después el alto.

```python
fig, ax = plt.subplots(figsize=(8, 4))
```

En este ejemplo, la figura tendrá 8 pulgadas de ancho y 4 pulgadas de alto.

### 6.5.5 Límites de los ejes

Podemos establecer los límites visibles de los ejes X e Y.

```python
ax.set_xlim(0, 10)
ax.set_ylim(0, 50)
```

- `set_xlim()`: establece los límites del eje X.
- `set_ylim()`: establece los límites del eje Y.

### 6.5.6 Rejilla

Podemos añadir una rejilla para facilitar la lectura y comparación de los valores.

```python
ax.grid(alpha=0.3, linestyle="--")
```

<img src="../docs/images/rejilla.png" alt="Grafico con rejilla" width="400">

- `ax.grid()`: añade una rejilla.
- `alpha`: controla la transparencia, con valores entre 0 y 1.
- `linestyle`: establece el estilo de las líneas de la rejilla.

También podemos utilizar:

```python
ax.grid(True)
```

Esto añade una rejilla con la configuración predeterminada.

### 6.5.7 Gráficos con varias series

Podemos representar varias series de datos en el mismo gráfico.

```python
fig, ax = plt.subplots()

ax.plot(dias, ventas, label="Ventas")
ax.plot(dias, visitas, label="Visitas")

ax.legend()
plt.show()
```

<img src="../docs/images/sin_legend.png" alt="Ejemplo grafico sin legend" width="400">
<img src="../docs/images/con_legend.png" alt="Ejemplo grafico con legend" width="400">

- `label`: asigna un nombre a cada serie de datos.
- `ax.legend()`: muestra la leyenda con esos nombres.

### 6.5.8 Formato de los ejes

Podemos evitar que Matplotlib muestre los valores del eje en notación científica.

```python
ax.ticklabel_format(style="plain", axis="y")
```
<img src="../docs/images/sin_notacion_cientifica.png" alt="Ejemplo grafico sin notacion científica" width="400">

Esto permite mostrar números completos en lugar de formatos como `5e3` o `1e6`.

### 6.5.9 Formatear fechas

Cuando el eje X contiene fechas, las etiquetas pueden solaparse. Para mejorar su legibilidad podemos utilizar:

```python
fig.autofmt_xdate()
```

<img src="../docs/images/fechas_con_autofmt.png" alt="Ejemplo grafico fechas con autofmt" width="400">

Esta función ajusta automáticamente la orientación de las etiquetas de fecha.

### 6.5.10 Guardar un gráfico

Podemos guardar una figura como archivo de imagen utilizando `savefig()`.

```python
fig.savefig(
    "grafico.png",
    dpi=300,
    bbox_inches="tight"
)
```

- `savefig()`: guarda la figura.
- `"grafico.png"`: nombre y formato del archivo.
- `dpi=300`: resolución de la imagen.
- `bbox_inches="tight"`: ajusta los márgenes para evitar espacios innecesarios.
