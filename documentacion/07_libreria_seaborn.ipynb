# 07. Librería Seaborn

## 7.1 ¿Qué es Seaborn?

Seaborn es una librería de Python construida sobre Matplotlib que facilita la creación de visualizaciones estadísticas y gráficas orientadas al análisis exploratorio de datos (EDA).

Permite crear determinados gráficos de forma más sencilla que Matplotlib, especialmente los relacionados con distribuciones, relaciones entre variables y comparación de categorías.

En Data Analytics se utiliza durante el EDA para detectar tendencias, distribuciones, diferencias entre grupos, relaciones y posibles valores atípicos.


## 7.2 Importación

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

`sns` es el alias habitual de Seaborn y `plt` el de `matplotlib.pyplot`. Seaborn utiliza Matplotlib internamente, por lo que ambas librerías suelen utilizarse conjuntamente.

> **Importante:** Seaborn no sustituye a Matplotlib. Seaborn facilita la creación de determinados gráficos estadísticos, mientras que Matplotlib proporciona un mayor control sobre la figura y su personalización.

## 7.4 Gráficos principales de Seaborn

### 7.4.1 `barplot()` — Gráfico de barras

Se utiliza para comparar una medida numérica entre distintas categorías. Por defecto, calcula la **media** de la variable numérica para cada categoría cuando existen varias observaciones.

```python
data = {
    "categoria": ["A", "A", "B", "B", "C", "C"],
    "precio": [10, 12, 8, 9, 14, 13],
}

df = pd.DataFrame(data)

sns.barplot(data=df, x="categoria", y="precio")
plt.title("Precio medio por categoría")
plt.show()
```

<img src="../docs/images_seaborn/precio_medio_por_categoria.png" alt="Ejemplo precio medio categoria" width="400">

Seaborn calcula el precio medio de cada categoría y lo representa mediante una barra.

**Cuándo utilizarlo:** para comparar medias u otras medidas estadísticas entre categorías, por ejemplo, el precio medio por producto o las ventas medias por país.

> Por defecto, `barplot()` también puede mostrar intervalos de incertidumbre alrededor de la estimación. No hay que confundirlos con la representación de todos los datos individuales.

### 7.4.2 `scatterplot()` — Gráfico de dispersión

Muestra la relación entre dos variables numéricas mediante puntos en un plano XY.

Permite explorar preguntas como:
- ¿A mayor precio hay menos ventas?
- ¿A más edad aumentan los ingresos?
- ¿Existe una relación entre inversión publicitaria y facturación?

```python
data = {
    "precio": [5, 8, 12, 15, 20, 25],
    "ventas": [200, 180, 160, 120, 100, 80],
}

df = pd.DataFrame(data)

sns.scatterplot(data=df, x="precio", y="ventas")
plt.title("Relación entre precio y ventas")
plt.show()
```

<img src="../docs/images_seaborn/relacion_entre_precio_ventas.png" alt="relacion entre precio y ventas" width="400">

En este ejemplo se observa una relación negativa: a medida que aumenta el precio, las ventas disminuyen. Es una relación visual en estos datos; por sí sola no demuestra causalidad.

### 7.4.3 `histplot()` — Histograma

Un histograma representa la distribución de una variable numérica. Divide sus valores en intervalos, llamados *bins*, y muestra cuántas observaciones caen dentro de cada intervalo.

```python
data = {
    "precio": [5, 8, 8, 10, 12, 14, 15, 16, 16, 16, 18, 20]
}

df = pd.DataFrame(data)

sns.histplot(data=df, x="precio", kde=True)
plt.title("Distribución de precios")
plt.show()
```

<img src="../docs/images_seaborn/distribucion_de_precios.png" alt="distribucion de precios" width="400">

- **Histograma:** las barras muestran la frecuencia de observaciones en cada intervalo.
- **KDE:** la curva representa una estimación suavizada de la distribución.

El parámetro `kde=True` añade esa curva al histograma.

### 7.4.4 `boxplot()` — Diagrama de caja

El diagrama de caja resume la distribución de una variable numérica y permite identificar posibles valores atípicos (*outliers*).

```python
data = {
    "horsepower": [210, 305, 212, 165, 102, 680]
}

df = pd.DataFrame(data)

sns.boxplot(data=df, y="horsepower")
plt.title("Distribución de potencia")
plt.show()
```

Elementos principales:

- **Caja:** representa el 50 % central de los datos, desde el primer cuartil (Q1, percentil 25) hasta el tercer cuartil (Q3, percentil 75).
- **Mediana:** línea dentro de la caja; divide los datos en dos mitades.
- **Bigotes:** llegan hasta los valores más alejados que siguen dentro del criterio establecido por el gráfico. Por defecto, Seaborn utiliza el criterio habitual de 1,5 veces el rango intercuartílico (IQR).
- **Outliers:** observaciones que quedan más allá de los bigotes y se muestran como puntos individuales.

El valor `680` del ejemplo puede aparecer como un posible valor atípico respecto al resto.

<img src="../docs/images_seaborn/boxplot.png" alt="boxplot" width="400">


> Un outlier no es necesariamente un dato erróneo. Hay que investigarlo y valorar el contexto antes de eliminarlo.

## 7.5 Gráficos complementarios de Seaborn

### 7.5.1 `kdeplot()` — Estimación de densidad

KDE (*Kernel Density Estimate*) representa de forma suavizada la distribución de una variable numérica mediante una curva.

Puede utilizarse como gráfico independiente o añadirse a un histograma.

```python
sns.kdeplot(data=df, x="precio", fill=True)
plt.title("Densidad de precios")
plt.show()
```

También podemos combinarlo con un histograma:

```python
sns.histplot(data=df, x="precio", kde=True)
plt.title("Distribución de precios")
plt.show()
```

Es habitual utilizar KDE junto con un histograma para visualizar simultáneamente las frecuencias y la forma suavizada de la distribución.

### 7.5.2 `violinplot()` — Diagrama de violín

Combina información de distribución y densidad. Permite comparar cómo se distribuyen los valores de una variable numérica entre diferentes categorías.

```python
import numpy as np

np.random.seed(0)

days = ["Thur", "Fri", "Sat", "Sun"]

data = {
    "day": np.repeat(days, 100),
    "total_bill": np.concatenate([
        np.random.normal(m, 5, 100)
        for m in [20, 25, 30, 27]
    ])
}

df = pd.DataFrame(data)

sns.violinplot(data=df, x="day", y="total_bill")
plt.title("Distribución del importe por día")
plt.show()
```

- **Anchura:** representa la densidad de los datos; las zonas más anchas concentran más observaciones.
- **Elemento central:** según la configuración, puede mostrar la mediana y otros cuartiles.

**Cuándo utilizarlo:** para comparar la forma y dispersión de una variable numérica entre varias categorías.

### 7.5.3 `pairplot()` — Análisis exploratorio multivariable

Genera una matriz de gráficos para explorar las relaciones entre varias variables numéricas.

- **Diagonal:** muestra la distribución individual de cada variable.
- **Fuera de la diagonal:** muestra gráficos de dispersión entre pares de variables.
- **`hue`:** permite colorear los puntos según una categoría.

**Cuándo utilizarlo:**
- Al principio de un análisis exploratorio (EDA).
- Para explorar visualmente relaciones entre varias variables.
- Cuando se trabaja con un número reducido de columnas numéricas; con demasiadas variables, la matriz puede resultar difícil de interpretar.

```python
df = sns.load_dataset("iris")

sns.pairplot(
    df,
    vars=["sepal_length", "sepal_width", "petal_length", "petal_width"],
    hue="species"
)

plt.show()
```

<img src="../docs/images_seaborn/pairplot.png" alt="pairplot" width="400"> 

> `sns.load_dataset("iris")` carga un conjunto de datos de ejemplo de Seaborn. La primera vez puede necesitar conexión a Internet. En un análisis real, normalmente utilizaríamos nuestro propio DataFrame.

### 7.5.4 `heatmap()` — Mapa de calor

Representa los valores de una tabla mediante colores. Se utiliza habitualmente para visualizar una matriz de correlación entre variables numéricas.

Permite detectar qué variables presentan una relación lineal positiva o negativa y qué pares pueden estar muy correlacionados.

```python
df = sns.load_dataset("iris")

corr = df[
    ["sepal_length", "sepal_width", "petal_length", "petal_width"]
].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Matriz de correlación")
plt.show()
```

- `corr()`: calcula la matriz de correlación. Por defecto utiliza la correlación de Pearson.
- `annot=True`: muestra los valores numéricos dentro de las celdas.
- `cmap="coolwarm"`: establece la paleta de colores.

**Interpretación de la correlación:**
- Valores cercanos a `1`: relación lineal positiva fuerte.
- Valores cercanos a `-1`: relación lineal negativa fuerte.
- Valores cercanos a `0`: relación lineal débil o inexistente.

> La correlación no implica causalidad. Además, una correlación cercana a cero no descarta que exista una relación no lineal entre las variables.

## 7.7 Seaborn en el trabajo de un Data Analyst

Seaborn resulta especialmente útil durante el análisis exploratorio de datos:

- `barplot()`: comparar métricas entre categorías.
- `scatterplot()`: explorar relaciones entre variables numéricas.
- `histplot()` y `kdeplot()`: analizar distribuciones.
- `boxplot()` y `violinplot()`: estudiar dispersión y posibles valores atípicos.
- `pairplot()`: explorar varias relaciones de forma conjunta.
- `heatmap()`: visualizar matrices de correlación.

La elección del gráfico depende de la pregunta que queremos responder y del tipo de variables que estamos analizando. No se trata de utilizar todos los gráficos, sino de seleccionar el que mejor ayude a entender los datos.
