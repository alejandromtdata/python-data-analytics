Introducción a Pandas para análisis de datos.

```python
import pandas as pd

ventas = {
    "cliente": ["Ana", "Carlos", "Laura", "Miguel", "Ana"],
    "producto": ["Monitor", "Teclado", "Monitor", "Ratón", "Portátil"],
    "importe": [500, 240, 250, 150, 1200]
}

df_ventas = pd.DataFrame(ventas)

df_ventas
```

Con type() podemos saber el tipo de dato con el que estamos trabajando. 

```python
print(type(df_ventas))
```

```python
df_ventas["importe"].sum()
```

Visualización de 2 columnas 

```python
df_ventas[["cliente", "importe"]]
```

si ponemos == nos muestra que registros de la columna "cliente" son igual a "Ana" con un bool True/False

```python
df_ventas["cliente"] == "Ana"
```

```python
df_ventas[df_ventas["cliente"] == "Ana"]
```

```python
df_ventas[df_ventas["cliente"] == "Carlos"]
```

```python
df_ventas[df_ventas["importe"] > 300]
```

```python
df_ventas[(df_ventas["cliente"] == "Carlos") | (df_ventas["cliente"] == "Ana") & (df_ventas["importe"] > 200)]
```

```python
df_ventas[((df_ventas["cliente"] == "Carlos") | (df_ventas["cliente"] == "Ana")) & (df_ventas["importe"] > 200)]
```

### Ordenar datos con sort_values()

En análisis de datos no basta con filtrar información: también necesitamos ordenarla para identificar los valores más altos, los más bajos o las operaciones más relevantes.

Por defecto, ordena de menor a mayor (ascending=True).

ascending=False indica que queremos un orden descendente.

```python
df_ventas.sort_values("importe", ascending=False)
```

```python
df_ventas.sort_values("importe").head(3)
```

```python
df_ventas.sort_values("importe", ascending=False).head(3)
```

```python
df_ventas[df_ventas["cliente"] == "Ana"]["cliente"]
```

```python
df_ventas["cliente"] == "Ana"
```

```python
df_ventas[df_ventas["cliente"] == "Ana"]
```

```python
df_ventas.groupby("producto")["importe"].sum().sort_values(ascending=False)
```
