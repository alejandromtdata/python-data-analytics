Que es una variable? 

Una variable es un nombre que utilizamos para almacenar y acceder a un valor durante la ejecución de un programa.


````python
ventas = 1500
````

Aquí tenemos tres elementos:

* ventas: nombre de la variable.
* =: operador de asignación.
* 1500: valor asignado.

Estamos almacenando el valor 1500 bajo el nombre ventas.

Podemos recuperar ese valor utilizando el nombre de la variable:

````python
ventas = 1500

print(ventas)
````

NOTA: en Python, = no significa igualdad matemática. Significa asignación.
Estamos asignando el valor de la derecha a la variable de la izquierda.

## 1.2 Modificar el valor de una variable

Una variable puede cambiar de valor durante la ejecución del programa.

````python
ventas = 1500

ventas = 2000

print(ventas)
````

Qué ha pasado?

Primero hemos asignado 1500 a ventas.
Después hemos sustituido ese valor por 2000.
Por tanto, cuando consultamos la variable, obtenemos su valor actual.

importante: no necesitamos introducir directamente los números en la operación.

Podemos utilizar los nombres de las variables.

Esto permite crear programas más dinámicos y fáciles de modificar.

### 2.6 Conversión de tipos de datos (Type Casting)

1. ¿Qué es la conversión de tipos?

La conversión de tipos consiste en transformar un dato de un tipo a otro.

Por ejemplo:

* Convertir un texto en número.
* Convertir un número entero en decimal.
* Convertir un número en texto.

¿Por qué necesitamos esto?

Imagina que COMPLEX SOLUTIONS recibe un archivo CSV donde la facturación aparece almacenada como texto.

Tenemos:

````python
facturacion = "25000"
````

Aunque visualmente parezca un número, Python lo interpreta como str.

Si intentamos realizar una operación matemática:

````python
facturacion = "25000"

resultado = facturacion + 5000
````

Obtendremos un error (TypeError), porque no podemos sumar directamente una cadena de texto y un número.

Necesitamos convertir el dato.

2. Funciones de conversión

Las principales funciones que utilizaremos son:

| Función | Conversión | Ejemplo |
|---|---|---|
| `int()` | Convierte a entero | `int("25")` → `25` |
| `float()` | Convierte a decimal | `float("19.99")` → `19.99` |
| `str()` | Convierte a texto | `str(150)` → `"150"` |
| `bool()` | Convierte a booleano | `bool(1)` → `True` |

3. Ejemplos prácticos

A. Conversión a entero

````python
unidades = "150"

unidades = int(unidades)

print(unidades)
print(type(unidades))
````

B. Conversión a decimal

````python
precio = "49.99"

precio = float(precio)

print(precio)
print(type(precio))
````

C. Conversión a texto

````python
id_cliente = 1045

id_cliente = str(id_cliente)

print(id_cliente)
print(type(id_cliente))
````

D. Conversión a booleano

````python
stock_disponible = 1

stock_disponible = bool(stock_disponible)

print(stock_disponible)
````

Una particularidad importante de bool():

| Valor original | Resultado |
|---|---|
| `0` | `False` |
| `1` | `True` |
| `""` | `False` |
| `"Hola"` | `True` |
| `None` | `False` |

NOTA: Cualquier cadena no vacía devuelve True, incluso "False".

### 2.7 Conversión de tipos: compatibilidad de valores

No todos los valores se pueden convertir a cualquier tipo de dato. Python comprueba si el contenido es compatible con la conversión solicitada.

````python
int("250")       # 250
int("hola")      # ValueError
float("25.8")    # 25.8
float("hola")    # ValueError
````

Importante: una cadena que contiene un número decimal no se puede convertir directamente a int.

````python
int("25.8")      # ValueError
````

Sin embargo, si el valor original ya es un número decimal (float), sí podemos convertirlo a entero:

````python
int(25.8)        # 25
````

### Diferencia entre int() y round()

Esta diferencia es importante porque ambas funciones se comportan de manera distinta.

| Función | Comportamiento | Ejemplo | Resultado |
|---|---|---|---|
| `int()` | Elimina la parte decimal, sin redondear. | `int(25.8)` | `25` |
| `round()` | Redondea al entero más cercano, siguiendo las reglas de redondeo de Python. | `round(25.8)` | `26` |

````python
int(25.9)       # 25
int(25.1)       # 25
int(-25.9)      # -25

round(25.8)     # 26
round(25.2)     # 25
````

Ojo: Python utiliza el redondeo al par en los empates exactos (por ejemplo, round(2.5) devuelve 2 y round(3.5) devuelve 4).

3. Errores durante la conversión: ValueError

ValueError es una excepción que aparece cuando una función recibe un valor cuyo contenido no es válido para realizar la operación solicitada.

Ejemplo:

````python
ventas = "dieciocho mil"
ventas = int(ventas)

# ValueError
````

En análisis de datos esto puede suceder cuando recibimos información de diferentes fuentes, como archivos CSV o Excel, y encontramos valores que no siguen el formato esperado.

Por eso, antes de convertir datos, debemos comprobar su contenido y validar que la conversión sea posible.


# 3. Operadores en Python

Los operadores son símbolos o palabras que permiten realizar operaciones con los datos y las variables.

En Python tenemos diferentes categorías de operadores. Vamos a estudiarlas progresivamente.

## 3.1 Operadores aritméticos

Son los que utilizamos para realizar operaciones matemáticas.

| Operador | Operación | Ejemplo | Resultado |
|---|---|---|---|
| `+` | Suma | `10 + 5` | `15` |
| `-` | Resta | `10 - 5` | `5` |
| `*` | Multiplicación | `10 * 5` | `50` |
| `/` | División | `10 / 5` | `2.0` |
| `//` | División entera | `10 // 3` | `3` |
| `%` | Módulo (resto) | `10 % 3` | `1` |
| `**` | Potencia | `2 ** 3` | `8` |

### 3.1.1 División /

Devuelve un resultado de tipo float, incluso cuando la división es exacta.

````python
resultado = 10 / 5

print(resultado)       # 2.0
print(type(resultado)) # float
````

### 3.1.2 División entera //

Devuelve el cociente entero de la división, descartando la parte decimal.

````python
resultado = 10 // 3

print(resultado) # 3
````

### 3.1.3 Módulo %

Devuelve el resto de una división.

````python
resultado = 10 % 3

print(resultado)
````

Este operador tiene aplicaciones prácticas. Por ejemplo, podemos utilizarlo para comprobar si un número es par o impar:

````python
numero = 10

print(numero % 2) 
````

Si el resto de dividir un número entre 2 es cero, significa que es par.

### 3.1.4 Potencia **

Permite elevar un número a una determinada potencia.

````python
resultado = 2 ** 3

print(resultado)
````

## 3.2 Operadores de comparación

Los operadores de comparación permiten comparar dos valores. El resultado de estas comparaciones siempre es un booleano: True o False.

Son fundamentales en Python porque los utilizaremos constantemente para establecer condiciones, filtrar información y tomar decisiones.

Tabla de operadores de comparación

| Operador | Significado | Ejemplo | Resultado |
|---|---|---|---|
| `==` | Igual que | `10 == 10` | `True` |
| `!=` | Distinto de | `10 != 5` | `True` |
| `>` | Mayor que | `10 > 5` | `True` |
| `<` | Menor que | `10 < 5` | `False` |
| `>=` | Mayor o igual que | `10 >= 10` | `True` |
| `<=` | Menor o igual que | `5 <= 10` | `True` |

Concepto importante: = frente a ==

No debemos confundirlos:

* = asigna un valor a una variable.
* == compara dos valores para comprobar si son iguales.

````python
ventas = 15000       # Asignación

resultado = ventas == 15000  # Comparación

print(resultado)  # True
````
